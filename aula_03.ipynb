```python
# 1. Operador and (E)

usuario_correto = "admin"
senha_correta = "12345"

usuario = input("Digite o usuário: ")
senha = input("Digite a senha: ")

if usuario == usuario_correto and senha == senha_correta:
  print("Login realizado com sucesso!")
else:
  print("Usuário ou senha incorretos!")
```

```python
# 1.1 and

idade = int(input("Digite a idade: "))

if idade >= 18 and idade <= 59:
  print("Seu votto é obrigatório!")

else:
  print("Nãoé obrigatório votar!")
```

```python
# 2. Operador or (ou)

dia_da_semana = input("Digite uma da semana:"). lower()

if dia_da_semana == "Sabado" or dia_da_semana == "Domingo" :
 print("Hoje é final de semana!")
else:
  print("Hoje é dia de trabalhar")
```

```python
# 3. operador not (não)

chovendo = False

if not chovendo:
  print("Vamos sair de casa!")
else:
  print("Não vamos sair de casa")
```
