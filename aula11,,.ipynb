```python
import pandas as pd
import io

csv_data = """Aluno,Portugues,Matematica
Alice,8,9
Bruno,6,7
Carla,7,8
Daniel,5,6
Eduarda,9,10
"""

df = pd.read_csv(io.StringIO(csv_data))
display(df)
```

```python
df['Media'] = (df['Portugues'] + df['Matematica']) / 2
display(df)
```

```python
alunos_aprovados = df[df['Media'] >= 7]
display(alunos_aprovados)
```
