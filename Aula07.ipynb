```python
# Desafio 1: Acima da Média

idades = [25, 45, 18, 30, 22, 60, 52]
soma_idades = sum(idades)
qtd_idades = len(idades)
media_idades = soma_idades / qtd_idades

print(f"Soma da idades: {soma_idades}")
print(f"Quantidade das idades: {qtd_idades}")
print(f"Média das idades: {media_idades:.2f}")

print("\nIdades acima da média:")
for idade in idades:
  if idade > media_idades:
    print(idade)
```
