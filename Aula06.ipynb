```python
# Desafio 13: Cálculo de Potência (sem função de potência)

base = int(input("Digite a base: "))
expoente = int(input("Digite o expoente: "))

resultado = 1

if expoente == 0:
    resultado = 1

elif expoente > 0:
    for _ in range(expoente):
        resultado *= base

else:
    for _ in range(abs(expoente)):
        resultado *= base
    resultado = 1 / resultado

print(f"O resultado de {base} elevado a {expoente} é: {resultado}")
```

```python
# Desafio 15: Gerador da Série de Fibonacci

while True:
    try:
        n = int(input("Digite o número de termos (N) para a série de Fibonacci: "))
        if n > 0:
            break
        else:
            print("Por favor, digite um número inteiro positivo.")
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

if n == 1:
    print("Série de Fibonacci (1 termo): 1")
elif n == 2:
    print("Série de Fibonacci (2 termos): 1, 1")
else:
    a, b = 1, 1
    fibonacci_series = [a, b]
    for _ in range(n - 2):
        a, b = b, a + b
        fibonacci_series.append(b)
    print(f"Série de Fibonacci ({n} termos): {', '.join(map(str, fibonacci_series))}")
```

```python
# Desafio 19: Menor, Maior e Soma de N Números

while True:
    try:
        n = int(input("Quantos números você deseja inserir? "))
        if n > 0:
            break
        else:
            print("Por favor, insira um número inteiro positivo.")
    except ValueError:
        print("Entrada inválida. Por favor, insira um número inteiro.")

soma = 0
maior = float('-inf') # Inicializa com o menor número possível
menor = float('inf')  # Inicializa com o maior número possível

print(f"\nPor favor, digite {n} números (entre 0 e 1000):")

for i in range(1, n + 1):
    while True:
        try:
            valor = float(input(f"Número {i}: "))
            if 0 <= valor <= 1000:
                break
            else:
                print("Número fora do intervalo. Por favor, digite um número entre 0 e 1000.")
        except ValueError:
            print("Entrada inválida. Por favor, digite um número válido.")

    soma += valor

    if valor > maior:
        maior = valor

    if valor < menor:
        menor = valor

print(f"\nO menor valor é: {menor}")
print(f"O maior valor é: {maior}")
print(f"A soma dos valores é: {soma}")
```

```python
# Desafio 23

N = int(input("Digite um número inteiro (N) para encontrar os primos até ele: "))

primes_found = []
division_count = 0


for num in range(2, N + 1):
    is_prime = True

    for i in range(2, num):
        division_count += 1
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        primes_found.append(num)

print(f"\nNúmeros primos entre 1 e {N}: {', '.join(map(str, primes_found))}")
print(f"Total de divisões executadas: {division_count}")
```

```python
# Desafio 24

while True:
    try:
        n_notas = int(input("Quantas notas você deseja inserir para calcular a média? "))
        if n_notas > 0:
            break
        else:
            print("Por favor, digite um número inteiro positivo para a quantidade de notas.")
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

soma_notas = 0

print(f"\nPor favor, digite as {n_notas} notas:")

for i in range(1, n_notas + 1):
    while True:
        try:
            nota = float(input(f"Nota {i}: "))

            break

        except ValueError:
            print("Entrada inválida. Por favor, digite um número para a nota.")
    soma_notas += nota

if n_notas > 0:
    media = soma_notas / n_notas
    print(f"\nA soma das notas é: {soma_notas:.2f}")
    print(f"A média aritmética das {n_notas} notas é: {media:.2f}")
else:
    print("Nenhuma nota inserida para calcular a média.")
```

```python
# Desafio 25

while True:
    try:
        num_pessoas = int(input("Quantas pessoas você deseja incluir na turma? "))
        if num_pessoas > 0:
            break
        else:
            print("Por favor, digite um número inteiro positivo para a quantidade de pessoas.")
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

soma_idades = 0

print(f"\nPor favor, digite a idade de cada uma das {num_pessoas} pessoas:")

for i in range(1, num_pessoas + 1):
    while True:
        try:
            idade = int(input(f"Idade da pessoa {i}: "))
            if idade >= 0:
                soma_idades += idade
                break
            else:
                print("A idade não pode ser negativa. Por favor, digite um valor válido.")
        except ValueError:
            print("Entrada inválida. Por favor, digite um número inteiro para a idade.")

if num_pessoas > 0:
    media_idade = soma_idades / num_pessoas
    print(f"\nA média de idade da turma é: {media_idade:.2f} anos.")

    if 0 <= media_idade <= 25:
        print("A turma é considerada JOVEM.")
    elif 26 <= media_idade <= 60:
        print("A turma é considerada ADULTA.")
    else:
        print("A turma é considerada IDOSA.")
else:
    print("Nenhuma pessoa inserida para calcular a média de idade.")
```

```python
# Desafio 26

votos_candidato_1 = 0
votos_candidato_2 = 0
votos_candidato_3 = 0

while True:
    try:
        total_eleitores = int(input("Digite o número total de eleitores: "))
        if total_eleitores > 0:
            break
        else:
            print("O número de eleitores deve ser positivo.")
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

print("\n--- Votação ---")
for i in range(1, total_eleitores + 1):
    while True:
        try:
            voto = int(input(f"Eleitor {i}: Digite o número do seu candidato (1, 2 ou 3): "))
            if voto == 1:
                votos_candidato_1 += 1
                break
            elif voto == 2:
                votos_candidato_2 += 1
                break
            elif voto == 3:
                votos_candidato_3 += 1
                break
            else:
                print("Voto inválido. Por favor, digite 1, 2 ou 3.")
        except ValueError:
            print("Entrada inválida. Por favor, digite um número inteiro para o voto.")

print("\n--- Resultado da Eleição ---")
print(f"Candidato 1: {votos_candidato_1} votos")
print(f"Candidato 2: {votos_candidato_2} votos")
print(f"Candidato 3: {votos_candidato_3} votos")
```

```python
# Desafio 27

while True:
    try:
        num_turmas = int(input("Digite a quantidade de turmas: "))
        if num_turmas > 0:
            break
        else:
            print("A quantidade de turmas deve ser um número inteiro positivo.")
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

soma_alunos = 0

print("\n--- Inserção de Alunos por Turma ---")
for i in range(1, num_turmas + 1):
    while True:
        try:
            num_alunos = int(input(f"Digite a quantidade de alunos para a turma {i}: "))
            if 0 <= num_alunos <= 40:
                soma_alunos += num_alunos
                break
            else:
                print("Número de alunos inválido. As turmas não podem ter mais de 40 alunos e não podem ter um número negativo de alunos.")
        except ValueError:
            print("Entrada inválida. Por favor, digite um número inteiro.")

if num_turmas > 0:
    media_alunos_por_turma = soma_alunos / num_turmas
    print(f"\n--- Resultado ---")
    print(f"O número médio de alunos por turma é: {media_alunos_por_turma:.2f}")
else:
    print("Nenhuma turma inserida.")
```
