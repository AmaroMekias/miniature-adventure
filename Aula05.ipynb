```python
# 1. O Laço while (Enquanto)

contador = 1

print("Vai começar a contagem!! ")

while contador <= 10:
 print(f"Contagem: {contador} ")
 contador = contador + 1

print("Terminou a contagem!!")
```

```python
# 2. O Laço for (Para) usando a função range ()

for i in range(1,8):
  print(f"Contagem de número {i} de 8")

print("------")

for i in range(2, 11, 2):
  print(f"Contagem de 2 em 2: {i}")
```
