```python
# Desafio 1: Função de Par ou Ímpar

def eh_par(numero):

  return numero % 2 == 0

try:
  numero_digitado = int(input("Digite um número para verificar se é par ou ímpar: "))


  if eh_par(numero_digitado):
    print(f"O número {numero_digitado} é Par.")
  else:
    print(f"O número {numero_digitado} é Ímpar.")
except ValueError:
  print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
# Desafio 2: Maior de Idade

def verifica_maioridade(idade):

  if idade >= 18:
    return "Maior de Idade"
  else:
    return "Menor de Idade"

# Pede a idade ao usuário
try:
  idade_digitada = int(input("Digite a idade para verificar a maioridade: "))

  # Chama a função e imprime o resultado
  resultado_maioridade = verifica_maioridade(idade_digitada)
  print(f"Com {idade_digitada} anos, a pessoa é: {resultado_maioridade}")
except ValueError:
  print("Entrada inválida. Por favor, digite um número inteiro para a idade.")
```
