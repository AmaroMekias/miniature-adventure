```python
nota = 8.5

print("Nota recebida:", nota)

if nota >= 9:
    situacao = "Excelente"
elif nota >= 7:
    situacao = "Bom"
elif nota >= 5:
    situacao = "Recuperação"
else:
    situacao = "Reprovado"

print("Situação final:", situacao)
```

```python
numeros = [2, 4, 6, 8]
soma = 0

print("Lista:", numeros)

for numero in numeros:
    print("Número atual:", numero)

    soma = soma + numero

    print("Soma parcial:", soma)

print("Soma final:", soma)
```

```python
def dobro(valor): # é uma função com nome dobro e um paremêtro"()"
    print("Valor recebido:", valor)
    resultado = valor * 2
    print("Resultado calculado:", resultado)
    return resultado

numero = 7
print("Número original:", numero)
resposta = dobro(numero)
print("Resposta final:", resposta)
```

```python
def contar_pares(lista):
    pares = 0

    for numero in lista:
        print("Número analisado:", numero)

        if numero % 2 == 0:
            pares = pares + 1
            print("Número par encontrado")

        print("Quantidade parcial de pares:", pares)

    return pares


valores = [3, 8, 5, 10, 12]

resultado = contar_pares(valores)

print("Total de pares:", resultado)
```
