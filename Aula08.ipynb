```python
# Desafio 1: Tabela de Preços

precos = {
    'maçã': 3.50,
    'banana': 2.00,
    'laranja': 1.80
}

produto_digitado = input('Digite o nome do produto: ')

if produto_digitado.lower() in precos:
    preco = precos[produto_digitado.lower()]
    print(f'O preço de {produto_digitado} é R${preco:.2f}')
else:
    print('Produto não encontrado.')
```
